```python
import numpy as np
import matplotlib.pyplot as plt

# Set up the plot style for a dark background
# plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(10, 7))

# Colors used in the drawing
blue_color = '#7eb0d5'
red_color = '#ff3b3b'
green_color = '#2ca02c'

# 1. Generate data for the Rate-Distortion curve (convex curve, e.g., y = 1/x)
x = np.linspace(0.15, 5, 200)
y = 1 / x

# Plot the blue curve
ax.plot(x, y, color=blue_color, linewidth=2)

# 2. Define points for the tangent lines
# Green tangent point (steeper slope)
x_g = 0.5
y_g = 1 / x_g
slope_g = -1 / (x_g**2)

# Red tangent point (shallower slope)
x_r = 1.5
y_r = 1 / x_r
slope_r = -1 / (x_r**2)

# Generate line data for tangents
x_line_g = np.linspace(0.1, 1.2, 100)
y_line_g = slope_g * (x_line_g - x_g) + y_g

x_line_r = np.linspace(0.1, 3.5, 100)
y_line_r = slope_r * (x_line_r - x_r) + y_r

# Plot the tangent lines
ax.plot(x_line_g, y_line_g, color=green_color, linewidth=1.5)
ax.plot(x_line_r, y_line_r, color=red_color, linewidth=1.5)

# 3. Plot dashed drop lines to the axes
# Green dashed lines
ax.plot([x_g, x_g], [0, y_g], color=green_color, linestyle='--', linewidth=1.5)
ax.plot([0, x_g], [y_g, y_g], color=green_color, linestyle='--', linewidth=1.5)

# Red dashed lines
ax.plot([x_r, x_r], [0, y_r], color=red_color, linestyle='--', linewidth=1.5)
ax.plot([0, x_r], [y_r, y_r], color=red_color, linestyle='--', linewidth=1.5)

# 4. Draw Custom Axes with Arrows
# Remove standard spines and ticks
for spine in ax.spines.values():
    spine.set_visible(False)
ax.set_xticks([])
ax.set_yticks([])

# Y-axis
ax.annotate('', xy=(0, 6), xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", color=blue_color, lw=1.5))
# X-axis
ax.annotate('', xy=(6, 0), xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", color=blue_color, lw=1.5))

# 5. Add Text Annotations
# Axis labels
# ax.text(-0.3, 5, 'Rate', color=blue_color, fontsize=16, rotation=90, verticalalignment='center')
# ax.text(5, -0.4, 'Distortion', color=blue_color, fontsize=16, horizontalalignment='center')

ax.set_ylabel('Rate', color=blue_color)
ax.

# Formula in the middle
ax.text(1.7, 2.5, r'$J = -D + \beta R$', color=red_color, fontsize=18)

# Text on the right
# ax.text(4.0, 3.5, r'$\beta : small :$', color=red_color, fontsize=14)
# ax.text(4.3, 3.1, 'infinite slope', color=red_color, fontsize=14)

# Set axis limits to frame it properly
# ax.set_xlim(-0.5, 6.5)
# ax.set_ylim(-0.5, 6.5)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

# --- Set standard plot formatting ---
plt.rcParams.update({
    'font.size': 12,
    'text.usetex': False,  # Simple text instead of LaTeX for speed
})

def plot_neat_distortion_curve():
    # Setup standard light theme figure
    fig, ax = plt.subplots(figsize=(10, 7))

    # Color palette suitable for a white background
    color_curve = '#1f77b4' # Standard blue
    color_green = '#2ca02c' # Standard green
    color_red   = '#d62728' # Standard red
    color_axis  = 'black'

    # --- 1. Data & Curve Generation ---
    # The curve looks like y = 1/x for simplicity
    x = np.linspace(0.15, 6, 200)
    y = 1 / x 
    ax.plot(x, y, color=color_curve, lw=2.5, zorder=2)

    # --- 2. Define Reference Points ---
    # Green Point: Closer to origin, steeper
    d_g = 0.5
    r_g = 1 / d_g
    
    # Red Point: Further out, shallower
    d_r = 1.5
    r_r = 1 / d_r

    # Draw specific points
    ax.scatter([d_g], [r_g], color=color_green, s=60, zorder=5)
    ax.scatter([d_r], [r_r], color=color_red, s=60, zorder=5)

    # Add text labels 'G' and 'R' slightly offset from the points
    ax.text(d_g + 0.1, r_g, 'G', color=color_green, fontsize=16, fontweight='bold', va='center')
    ax.text(d_r + 0.1, r_r, 'R', color=color_red, fontsize=16, fontweight='bold', va='center')


    # --- 3. Compute and Draw Tangent Lines (Slopes) ---
    # We use calculus: slope of 1/x is -1/(x^2)
    slope_g = -1 / (d_g**2) # Steep negative
    slope_r = -1 / (d_r**2) # Shallower negative

    # Generate tangent lines data
    t_range = np.linspace(0.01, 7, 300)
    tangent_g = slope_g * (t_range - d_g) + r_g
    tangent_r = slope_r * (t_range - d_r) + r_r

    ax.plot(t_range, tangent_g, color=color_green, lw=1.5, ls='-', alpha=0.8, zorder=1)
    ax.plot(t_range, tangent_r, color=color_red, lw=1.5, ls='-', alpha=0.8, zorder=1)


    # --- 4. Draw Neat Perpendicular Lines (Projections to axes) ---
    # Use dotted lines matching their respective tangent colors
    # From G
    ax.plot([d_g, d_g], [0, r_g], color=color_green, ls=':', lw=1.5, zorder=1)
    ax.plot([0, d_g], [r_g, r_g], color=color_green, ls=':', lw=1.5, zorder=1)
    
    # From R
    ax.plot([d_r, d_r], [0, r_r], color=color_red, ls=':', lw=1.5, zorder=1)
    ax.plot([0, d_r], [r_r, r_r], color=color_red, ls=':', lw=1.5, zorder=1)


    # --- 5. Clean Custom Axes & Labels ---
    # Hide standard ticks and box
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values(): spine.set_visible(False)

    # Use ArrowPatch for axes arrows
    y_arrow = FancyArrowPatch((0, 0), (0, 7), arrowstyle='-|>', color=color_axis, mutation_scale=20, lw=1.5)
    x_arrow = FancyArrowPatch((0, 0), (7.5, 0), arrowstyle='-|>', color=color_axis, mutation_scale=20, lw=1.5)
    ax.add_patch(y_arrow)
    ax.add_patch(x_arrow)

    # Axis Labels
    ax.text(-0.4, 6.5, 'Rate', color=color_axis, rotation=90, fontsize=16, va='center', ha='right')
    ax.text(7.2, -0.3, 'Distortion', color=color_axis, fontsize=16, ha='right')

    # Specific axis point labels (G & R values)
    ax.text(d_g, -0.4, r'$D_G$', color='black', ha='center', va='top', fontsize=14)
    ax.text(d_r, -0.4, r'$D_R$', color='black', ha='center', va='top', fontsize=14)
    ax.text(-0.1, r_g, r'$R_G$', color='black', ha='right', va='center', fontsize=14)
    ax.text(-0.1, r_r, r'$R_R$', color='black', ha='right', va='center', fontsize=14)


    # --- 6. Add Neat Formulas ---
    # Central Formula placed near the intersection point R
    ax.text(3, 2.2, r'$J = D + \beta R$', color=color_red, fontsize=20, fontweight='bold', zorder=10)
    ax.text(5.1, 2.2, r'(Slope = $-\frac{1}{\beta}$)', color=color_red, fontsize=14, va='center', zorder=10)


    # --- 7. Final Adjustments ---
    ax.set_xlim(-0.7, 8) 
    ax.set_ylim(-0.8, 7.5)
    ax.set_aspect('equal') # Keep the geometry correct

    plt.tight_layout()
    plt.show()

# Run the plotting function
plot_neat_distortion_curve()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.family": "serif",
    "mathtext.fontset": "dejavuserif",
})

fig, ax = plt.subplots(figsize=(9, 6.8))
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# ── R(D) = log(σ/D) ───────────────────────────────────────────────────────────
sigma = 1.0
R_top = 3.2
D_start = sigma * np.exp(-R_top)
D = np.linspace(D_start, sigma, 700)
R = np.log(sigma / D)
ax.plot(D, R, color="#3B82F6", linewidth=2.2, zorder=3)

# ── Tangent points ────────────────────────────────────────────────────────────
beta_g = 6.0;  D_t_g = 1.0/beta_g;  R_t_g = np.log(beta_g)
beta_r = 2.5;  D_t_r = 1.0/beta_r;  R_t_r = np.log(beta_r)

def tline(D_t, R_t, beta, Dvals):
    return R_t - beta * (Dvals - D_t)

x_max = sigma * 1.22
y_max = R_top * 1.06

D_g = np.linspace(D_t_g - R_t_g/beta_g - 0.02, D_t_g + (R_t_g + 0.3)/beta_g, 300)
R_g = tline(D_t_g, R_t_g, beta_g, D_g)
mask_g = (R_g >= -0.3) & (R_g <= y_max) & (D_g >= -0.04) & (D_g <= x_max)
ax.plot(D_g[mask_g], R_g[mask_g], color="#16A34A", linewidth=1.9, zorder=2)

D_r = np.linspace(0.0, sigma * 1.05, 300)
R_r = tline(D_t_r, R_t_r, beta_r, D_r)
mask_r = (R_r >= -0.05) & (R_r <= y_max) & (D_r >= -0.04) & (D_r <= x_max)
ax.plot(D_r[mask_r], R_r[mask_r], color="#DC2626", linewidth=1.9, zorder=2)

# ── Dashed projections ────────────────────────────────────────────────────────
dk_g = dict(color="#16A34A", linewidth=1.0, linestyle="--", zorder=1)
dk_r = dict(color="#DC2626", linewidth=1.0, linestyle="--", zorder=1)

ax.plot([0, D_t_g], [R_t_g, R_t_g], **dk_g)
ax.plot([D_t_g, D_t_g], [0, R_t_g], **dk_g)
ax.plot([0, D_t_r], [R_t_r, R_t_r], **dk_g)
ax.plot([D_t_r, D_t_r], [0, R_t_r], **dk_r)

# ── Axes limits ───────────────────────────────────────────────────────────────
ax.set_xlim(-0.12, x_max)
ax.set_ylim(-0.50, y_max)

# ── Axes arrows ───────────────────────────────────────────────────────────────
ax.annotate("", xy=(x_max * 0.988, 0), xytext=(-0.08, 0),
            arrowprops=dict(arrowstyle="-|>", color="black", lw=1.3))
ax.annotate("", xy=(0, y_max * 0.980), xytext=(0, -0.42),
            arrowprops=dict(arrowstyle="-|>", color="black", lw=1.3))

# ── Axis titles ───────────────────────────────────────────────────────────────
ax.text(x_max * 0.992, -0.16, r"Distortion, $D$", fontsize=12,
        ha="right", va="top", style="italic")
ax.text(0.020, y_max * 0.978, r"Rate, $R$", fontsize=12,
        ha="left", va="top", style="italic")

# ── x-axis tick marks & labels ────────────────────────────────────────────────
tk = 0.045   # tick half-height
for D_val, lbl in [(D_t_g, r"$D^*_{\!\beta_1}$"),
                   (D_t_r, r"$D^*_{\!\beta_2}$"),
                   (sigma,  r"$\sigma$")]:
    ax.plot([D_val, D_val], [-tk, tk], color="black", linewidth=1.1)
    ax.text(D_val, -tk - 0.06, lbl, fontsize=10.5, ha="center", va="top")

# ── y-axis tick marks & labels ────────────────────────────────────────────────
tkx = 0.012
for R_val, lbl in [(R_t_g, r"$R^*_{\!\beta_1}$"),
                   (R_t_r, r"$R^*_{\!\beta_2}$")]:
    ax.plot([-tkx, tkx], [R_val, R_val], color="black", linewidth=1.1)
    ax.text(-tkx - 0.015, R_val, lbl, fontsize=10.5, ha="right", va="center")

# Origin
ax.text(-0.015, -0.12, r"$0$", fontsize=10.5, ha="right", va="top")

# ── Slope labels — placed along each line, away from clutter ─────────────────
# Green: label in the upper portion of the green line
D_lg = 0.04
R_lg = tline(D_t_g, R_t_g, beta_g, D_lg)
ax.text(D_lg + 0.025, R_lg - 0.12, r"slope $= -\beta_1$",
        fontsize=9.5, color="#16A34A", ha="left", va="top",
        rotation=-72, rotation_mode="anchor")

# Red: label in right portion of red line, above the blue curve
D_lr = 0.72
R_lr = tline(D_t_r, R_t_r, beta_r, D_lr)
ax.text(D_lr, R_lr + 0.12, r"slope $= -\beta_2$",
        fontsize=9.5, color="#DC2626", ha="left", va="bottom",
        rotation=-35, rotation_mode="anchor")

# ── R(D) curve label — upper right of curve ───────────────────────────────────
D_cl = 0.78
R_cl = np.log(sigma / D_cl)
ax.text(D_cl + 0.03, R_cl + 0.10, r"$R(D) = \log\!\left(\frac{\sigma}{D}\right)$",
        fontsize=10.5, color="#3B82F6", ha="left", va="bottom")

# ── J label ───────────────────────────────────────────────────────────────────
ax.text(0.33, 1.62, r"$J = -D + \beta R$",
        fontsize=12, color="#DC2626", ha="left")

for spine in ax.spines.values():
    spine.set_visible(False)
ax.set_xticks([])
ax.set_yticks([])

plt.tight_layout(pad=1.2)
# plt.savefig("/mnt/user-data/outputs/rate_distortion_labelled.png",
            # dpi=180, bbox_inches="tight", facecolor="white")
# print("Done")
plt.show()

```
